```python
from textblob import TextBlob
import matplotlib.pyplot as plt

# Emoji Helper
def sentiment_emoji(score):
    if score > 0:
        return " Positive"
    elif score < 0:
        return " Negative"
    else:
        return " Neutral"
```

```python
# Input
text = input("Enter a sentence or paragraph for sentiment analysis:\n")

# Create TextBlob
blob = TextBlob(text)

# Sentiment Scores
polarity = blob.sentiment.polarity
subjectivity = blob.sentiment.subjectivity

# Results
print("\n Sentiment Analysis Result:")
print(f" Polarity: {polarity}")
print(f" Subjectivity: {subjectivity}")
print(f" Sentiment: {sentiment_emoji(polarity)}")
```
