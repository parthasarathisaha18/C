```python
import csv
import os

# File path (in Colab, this saves in current runtime)
FILENAME = 'student_marks.csv'
students = {}

# 🔹 Load data from CSV on start
def load_data():
    if os.path.exists(FILENAME):
        with open(FILENAME, mode='r') as file:
            reader = csv.reader(file)
            next(reader) # Skip the header row
            for row in reader:
                if row:
                    students[row[0]] = int(row[1])
        print(f"✅ Loaded {len(students)} student records.\n")
    else:
        print("⚠️ No existing data found.\n")

# 🔹 Save data to CSV
def save_data():
    with open(FILENAME, mode='w') as file:
        writer = csv.writer(file)
        writer.writerow(["Name", "Marks"]) # Write the header row
        for name, marks in students.items():
            writer.writerow([name, marks])

# 🔹 Add student
def add_student():
    name = input("Enter student name: ")
    marks = int(input("Enter marks: "))
    students[name] = marks
    save_data()
    print(f"{name} added successfully!\n")

# 🔹 Show topper
def get_topper():
    if students:
        topper = max(students.items(), key=lambda x: x[1])
        print(f"🏆 Topper: {topper[0]} with {topper[1]} marks\n")
    else:
        print("No students added yet!\n")

# 🔹 Sort and show list
def sort_students():
    if students:
        sorted_list = sorted(students.items(), key=lambda x: x[1], reverse=True)
        print("📊 Sorted Student List:")
        for i, (name, marks) in enumerate(sorted_list, 1):
            print(f"{i}. {name}: {marks}")
        print()
    else:
        print("No students added yet!\n")

# 🔹 Display Menu
def display_menu():
    print("🎓 Student Marks Manager")
    print("1. Add Student")
    print("2. Show Topper")
    print("3. Show Sorted List")
    print("4. Exit")

# 🔹 Start Program
load_data()
while True:
    display_menu()
    choice = input("Choose an option (1-4): ")

    if choice == '1':
        add_student()
    elif choice == '2':
        get_topper()
    elif choice == '3':
        sort_students()
    elif choice == '4':
        print("Exiting program. Goodbye!")
        break
    else:
        print("Invalid option. Try again.\n")
```
